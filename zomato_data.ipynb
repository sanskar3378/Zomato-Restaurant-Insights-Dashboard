```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Zomato_data.csv')
```

```python
df.info()
```

```python
df.head()
```

```python
# Conveerting rate column
''' removing the /5 text from all the columns and converting the remaining element to float '''

df['rate'] = df['rate'].str.removesuffix(suffix='/5')
df['rate'] = df['rate'].astype(float)
```

```python
df.head()
```

```python
''' Checking for NULL values '''

df.isnull().sum()
#  0 confirms that there is are no NULL values in the dataset
```

```python
df.info()
```

```python
'''Converting cost column to float'''
df['approx_cost(for two people)'] = df['approx_cost(for two people)'].astype(float)
```
